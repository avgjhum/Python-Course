### Упражнение 3
Напишите программу на Python, которая запрашивает у пользователя два числа и выводит их сумму.

```python
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
print("Sum of " + str(num1) + " and " + str(num2) + " is " + str(num1 + num2))
```

### Упражнение 4
Напишите программу на Python, которая запрашивает у пользователя три слова и выводит их в обратном порядке, разделяя пробелами.

```python
print("Hello! Can you enter 3 words? :>")
word1 = str(input("Enter first word: "))
word2 = str(input("Enter second word: "))
word3 = str(input("Enter third word: "))
print("Your 3 words: " + word1 + " " + word2 + " " + word3) #Выводит слова по порядку
print("Your 3 words from the end: " + word3 + " " + word2 + " " + word1) #Выводит их же в обратном порядке
```
